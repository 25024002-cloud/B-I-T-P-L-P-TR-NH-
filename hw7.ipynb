W7A19 Merge 2 mảng đã sắp xếp (có thể tăng dần, có thể giảm dần) mà không
dùng bất cứ thuật toán sắp xếp nào

```python
L1 = list(map(int, input().split()))
L2 = list(map(int, input().split()))
if len(L1) > 1 and L1[0] > L1[-1]:
    L1 = L1[::-1]
if len(L2) > 1 and L2[0] > L2[-1]:
    L2 = L2[::-1]
res = []
i = 0  # Vị trí đang xét của mảng 1
j = 0  # Vị trí đang xét của mảng 2
while i < len(L1) and j < len(L2):
    if L1[i] < L2[j]:
        res.append(L1[i])
        i = i + 1  # Chọn xong thì nhích vị trí mảng 1 lên
    else:
        res.append(L2[j])
        j = j + 1  # Chọn xong thì nhích vị trí mảng 2 lên
# 3. Vét nốt: Nếu một mảng hết rồi, mảng kia còn ai thì hốt hết vào
while i < len(L1):
    res.append(L1[i])
    i = i + 1
while j < len(L2):
    res.append(L2[j])
    j = j + 1
print(*(res))
```

W7A25 Tìm số có đúng k chữ số lớn nhất có thể trong một số có n chữ số (coi
đầu vào là một số rất lớn - str). Không được hoán đổi vị trí các chữ số

```python
num_str = input().strip()
k = int(input())
n = len(num_str)
result = []
start_idx = 0
# Cậu cần tìm đúng k chữ số, nên ta chạy vòng lặp k lần
for i in range(k):
    # Giới hạn khoảng được phép tìm số lớn nhất để không bị thiếu số phía sau
    end_idx = n - (k - 1 - i)
    # Cắt chuỗi trong khoảng an toàn đó
    search_range = num_str[start_idx:end_idx]
    # Tìm chữ số lớn nhất trong khoảng này
    max_char = max(search_range)
    result.append(max_char)
    # Cập nhật lại vị trí bắt đầu cho lần tìm kiếm tiếp theo (bỏ qua các số đứng trước)
    start_idx = start_idx + search_range.find(max_char) + 1
# In kết quả cuối cùng
print("".join(result))
```

W7A28 Bà̀i toán Tìm dãy các hợp số dài nhất liên tiếp không chứa số nguyên
nào <= N (N là số nguyên dương cho trước). Nếu có nhiều dãy cùng độ
dài thì lấy dãy nhỏ nhất (gặp đầu tiên từ trái qua)

```python
import math
n = int(input())
if n <= 1:
    print([])
else:
    # Tính N giai thừa (N!)
    giai_thua = math.factorial(n)
    # Tạo dãy các hợp số liên tiếp: từ (N! + 2) đến (N! + N)
    kqua = []
    for i in range(2, n + 1):
        kqua.append(giai_thua + i)        
    # In ra kết quả đúng định dạng list Python theo yêu cầu
    print(kqua)
```

W7A29 Tìm Dãy con tăng dần liên tục dài nhất của một danh sách nhập vào từ
bàn phím. Lấy dãy đầu tiên nếu có nhiều dãy có số phần tử bằng nhau

```python
lst = list(map(int, input().split()))

if not lst:
    print([])
else:
    day_max = [lst[0]]      # Dãy con dài nhất tìm được từ trước đến nay
    day_hien_tai = [lst[0]] # Dãy con đang được nuôi/đếm
    for i in range(1, len(lst)):
        if lst[i] > lst[i-1]:
            # Số sau lớn hơn số trước thì ném thẳng số đó vào dãy hiện tại
            day_hien_tai.append(lst[i])
        else:
            # Bị ngắt quãng! So sánh độ dài trước khi reset
            if len(day_hien_tai) > len(day_max):
                day_max = day_hien_tai
            # Tạo một dãy mới bắt đầu bằng chính số hiện tại
            day_hien_tai = [lst[i]]
    # CHECK LẦN CUỐI: Phòng trường hợp dãy tăng nằm ở cuối mảng
    if len(day_hien_tai) > len(day_max):
        day_max = day_hien_tai
    print(day_max)

```

W7A30 Tìm Dãy con tăng dần KHÔNG liên tục dài nhất của một danh sách nhập
vào từ bàn phím

```python
lst = list(map(int, input().split()))
if not lst:
    print(0)
else:
    n = len(lst)
    # Khởi tạo mảng dp có n phần tử, tất cả ban đầu đều bằng 1
    # Vì bản thân mỗi số đứng một mình đã là một dãy tăng có độ dài là 1 rồi
    dp = [1] * n
    
    # Vòng lặp i chạy từ đầu đến cuối mảng để xét từng số một
    for i in range(0, n):
        # Vòng lặp j chạy quét qua tất cả các số đứng TRƯỚC số i
        for j in range(0, i):
            # Nếu tìm thấy một số arr[j] nhỏ hơn arr[i] (thỏa mãn luật tăng dần)
            if lst[i] > [j]:
                # Cập nhật độ dài lớn nhất có thể đạt được tại vị trí i
                dp[i] = max(dp[i], dp[j] + 1)
                
    # Kết quả cuối cùng là độ dài lớn nhất tìm được trong mảng dp
    print(max(dp))
```
