W7A1: Viết hàm tìm kiếm nhị phân
trên một danh sách đã
được sắp xếp tăng dần.
Trả về chỉ số của phần tử
hoặc -1 nếu không tìm
thấy. mẫu hàm: def
binary_search(arr, left,
right, target)

```python
def binary_search(arr, left, right, target):
    if left > right: #nếu 0 có khoảng nào ktra
        return -1
    mid = (left+right)//2
    if arr[mid] == target: #xem giữa ph mục tiêu k
        return mid
    elif arr[mid] < target: # mục tiêu lớn hơn mid
        return binary_search(arr, mid+1, right, target)
    else:
        return binary_search(arr, left, mid-1, target)

```

W7A2:Cho danh sách arr và một
số nguyên x, đếm xem x
xuất hiện bao nhiêu lần
trong danh sách. def
count_occurrences(arr, x) ->int:

```python
def count_occurrences(arr, x)->int :
    dem = 0
    for i in arr:
        if i == x:
            dem = dem +1
    return dem
```

W7A3 Viết hàm thực hiện thuật
toán bubble sort, Selection
Sort để sắp xếp danh sách
theo thứ tự tăng dần. Sắp
xếp nổi bọt (Bubble Sort); Sắp xếp chọn (Selection
Sort)

```python
def bubble_sort(arr): # đẩy lớn nhất xuống cuối
    n = len(arr)
    for i in range(n):
        swapped = False
        for j in range(0, n-i-1):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
                swapped = True
        if not swapped:
            break
    return arr
```

```python
def selection_sort(arr): # đẩy nhỏ nhất lên đầu
    n = len(arr)
    for i in range(n):
        min_idx = i
        for j in range(i+1, n):
            if arr[j] < arr[min_idx]:
                min_idx = j
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
    return arr
```

W7A5 Cho danh sách các số
nguyên và một số X, đếm
tất cả các cặp (a, b) sao
cho a + b = X. c

```python
def dem_cap_chia_het_X(arr, X):
    n = len(arr)
    dem = 0
    for i in range(n):
        for j in range(i+1, n):
            if (arr[i]+arr[j])%X==0:
                dem +=1
    return dem
```

W7A8: Bạn được cung cấp: một
danh sách các khoảng thời
gian interval = [[l1, r1], [l2,
r2], ...] và một danh sách
các truy vấn queries = [q1,
q2, ...]
Với mỗi truy vấn q, hãy tìm
độ dài nhỏ nhất của
khoảng thời gian [li, ri] sao
cho li ≤ q ≤ ri.
Nếu không tồn tại khoảng
thời gian nào như vậy, hãy
trả về -1 cho truy vấn đó.
solve in < O(m.n), hint:
binary search in the
queries list

```python
import heapq, sys

# Đọc toàn bộ dữ liệu một lần cho nhanh
data = sys.stdin.read().split()
n, m = int(data[0]), int(data[1])
intervals = sorted([[int(data[i]), int(data[i+1])] for i in range(2, 2*n+2, 2)])
queries = sorted([(int(q), idx) for idx, q in enumerate(data[2*n+2:])])

res, heap, i = [-1] * m, [], 0

for q, idx in queries:
    # 1. Ném tất cả hộp có thể chứa q vào giỏ (heap)
    while i < n and intervals[i][0] <= q:
        l, r = intervals[i]
        heapq.heappush(heap, (r - l + 1, r))
        i += 1
    
    # 2. Vứt những hộp đã quá hạn (kết thúc trước q)
    while heap and heap[0][1] < q:
        heapq.heappop(heap)
    
    # 3. Lấy thằng ngắn nhất ở đầu giỏ
    if heap:
        res[idx] = heap[0][0]

print(*(res))
```

W7A10 Cho một danh sách các số

nguyên m, hãy tìm tổng
mảng con lớn nhất theo
modulo m.
Tức là, với bất kỳ mảng
con nums[i:j] nào, hãy tính
sum(nums[i:j]) % m và trả
về kết quả lớn nhất có thể.

```python
nums= list(map(int, input().split()))
m = int(input())
n = len(nums)
max_modulo = 0
for i in range(n):
    tong_tam = 0
    for j in range(i, n):
        tong_tam = (tong_tam +nums[j])%m
        if tong_tam > max_modulo:
            max_modulo = tong_tam
print(max_modulo)
        
```

W7A13 Dãy hài hòa là một dãy
thỏa mãn điều kiện giá trị
lớn nhất và giá trị nhỏ nhất
của dãy khác nhau đúng
bằng \( 1 \).
Viết hàm trả về độ dài của
dãy hài hòa con dài nhất có
thể có trong mảng giá trị
nguyên \( arr \) có \( n \)
phần tử
Ví dụ:
Input: 1 3 2 2 5 2 3 7
Output: 5 - Dãy hài hòa con
dài nhất là [3,2,2,2,3]

```python
arr = list(map(int, input().split()))
max_length = 0
# Duyệt qua từng số trong mảng
for x in arr:
    # 1. Đếm xem có bao nhiêu số x
    count_x = arr.count(x)
    # 2. Đếm xem có bao nhiêu số x + 1
    count_x_plus_1 = arr.count(x + 1) 
    # 3. Nếu thực sự có thằng hàng xóm (count > 0)
    if count_x_plus_1 > 0:
        # Tổng độ dài = số lượng x + số lượng x+1
        current_length = count_x + count_x_plus_1    
        # 4. Cập nhật kỷ lục
        if current_length > max_length:
            max_length = current_length

print(max_length)
```
