W9A16 Cho hai số nguyên dương, n và m (0 < n, m <= 20 ). .Bạn hãy in ra tất cả các chuỗi nhị phân có
n số 0 và m số 1. In theo thứ tự tăng dần trong từ điển

```python
n, m = map(int, input().split())

def backtrack(zeros, ones, current_str):
    if zeros == 0 and ones == 0:
        print(current_str)
        return
    
    if zeros > 0:
        backtrack(zeros - 1, ones, current_str + "0")
    if ones > 0:
        backtrack(zeros, ones - 1, current_str + "1")

backtrack(n, m, "")
```

W9A17 Viết hàm tìm dãy con tăng nghiêm ngặt dài nhất trong dãy đầu vào từ bàn phím 


```python
def find_lis(arr):
    if not arr:
        return []
    
    dp = [1] * len(arr)
    parent = [-1] * len(arr)
    
    for i in range(1, len(arr)):
        for j in range(i):
            if arr[i] > arr[j] and dp[i] < dp[j] + 1:
                dp[i] = dp[j] + 1
                parent[i] = j
                
    max_len = max(dp)
    idx = dp.index(max_len)
    
    res = []
    while idx != -1:
        res.append(arr[idx])
        idx = parent[idx]
        
    res.reverse()
    return res

arr = list(map(int, input().split()))
print(*find_lis(arr))
```

W9A18 Cho dãy N phần tử số nguyên, hãy đếm số cặp có vị trí khác nhau và có tổng bằng k (số nguyên
dương cho vào từ bàn phím)


```python
arr = list(map(int, input().split()))
k = int(input())

count = 0
seen = {}

for x in arr:
    target = k - x
    if target in seen:
        count += seen[target]
    seen[x] = seen.get(x, 0) + 1

print(count)
```

W9A20 Thực hiện chuẩn hóa chuỗi kí tự:
a. bỏ khoảng trắng đầu cuối
b. các khoảng trắng giữa chỉ còn 1 dấu cách
c. viết hoa chữ cái đầu mỗi từ
KHÔNG DÙNG CÁC HÀM SẴN CỐ

```python
s = input()
res = ""

start = 0
while start < len(s) and s[start] == ' ':
    start += 1
    
end = len(s) - 1
while end >= 0 and s[end] == ' ':
    end -= 1

new_word = True
for i in range(start, end + 1):
    c = s[i]
    if c == ' ':
        if not new_word:
            res += ' '
            new_word = True
    else:
        if new_word:
            if 'a' <= c <= 'z':
                res += chr(ord(c) - 32)
            else:
                res += c
            new_word = False
        else:
            if 'A' <= c <= 'Z':
                res += chr(ord(c) + 32)
            else:
                res += c

print(res)
```

W9A21 Tìm phần tử lớn thứ k trong dãy n phần tử bằng thuật toán tốt nhất có thể – Tránh dùng sắp xếp
(dùng phân hoạch của quicksort)

```python
def quickselect(arr, k):
    if not arr:
        return None
        
    pivot = arr[len(arr) // 2]
    left = [x for x in arr if x > pivot]
    mid = [x for x in arr if x == pivot]
    right = [x for x in arr if x < pivot]
    
    L = len(left)
    M = len(mid)
    
    if k <= L:
        return quickselect(left, k)
    elif k > L + M:
        return quickselect(right, k - L - M)
    else:
        return pivot

arr = list(map(int, input().split()))
k = int(input())
print(quickselect(arr, k))
```

W9A22 Hai chuỗi là anagram nếu chúng có cùng ký tự và số lần xuất hiện. viết CT kiểm tra xem 2 chuỗi
đầu vào có phải là anagram của nhau không 


```python
s1 = input()
s2 = input()

def is_anagram(str1, str2):
    if len(str1) != len(str2):
        return False
        
    count = {}
    for c in str1:
        count[c] = count.get(c, 0) + 1
        
    for c in str2:
        if count.get(c, 0) == 0:
            return False
        count[c] -= 1
        
    return True

print(is_anagram(s1, s2))
```

W9A23 Cho số nguyên rất lớn (string), và một số nguyên dương k. Tìm số có k chữ số lớn nhất có thể
từ số ban đầu bằng cách xóa bớt đi một số chữ số nào đó mà không thay đổi vị trí của bất kì
chữ số nào. (VD: 78934340999, k = 7 => 9440999 là số lớn nhất có 7 chữ số cần tìm)

```python
s = input().strip()
k = int(input())

stack = []
can_remove = len(s) - k

for digit in s:
    while stack and can_remove > 0 and stack[-1] < digit:
        stack.pop()
        can_remove -= 1
    stack.append(digit)

print("".join(stack[:k]))
```

W9A24 Nén chuối kí tự bằng cách đếm số lần xuất hiện của các kí tự liên tiếp nhau và ghi lại ngay sau
kí tự đó. VD aaaaaafdhhhhh =>a6f1d1h5

```python
s = input()
res = ""

if s:
    count = 1
    curr = s[0]
    for i in range(1, len(s)):
        if s[i] == curr:
            count += 1
        else:
            res += curr + str(count)
            curr = s[i]
            count = 1
    res += curr + str(count)

print(res)
```

W9A25 Giả sử mật khẩu mạnh là một chuỗi kí tự dài ít nhất 8 kí tự gồm ít nhất 1 chữ cái thường, 1 chữ
cái hoa, 1 chữ số và 1 trong các kí tự đặc biệt !, @, #, $. Nhập vào một chuỗi kĩ tự, kiểm tra xem
nó có phải là mật khẩu mạnh không. 

```python
s = input()

if len(s) >= 8:
    has_lower = False
    has_upper = False
    has_digit = False
    has_special = False
    
    for c in s:
        if 'a' <= c <= 'z':
            has_lower = True
        elif 'A' <= c <= 'Z':
            has_upper = True
        elif '0' <= c <= '9':
            has_digit = True
        elif c in "!@#$":
            has_special = True
            
    if has_lower and has_upper and has_digit and has_special:
        print("True")
    else:
        print("False")
else:
    print("False")
```
