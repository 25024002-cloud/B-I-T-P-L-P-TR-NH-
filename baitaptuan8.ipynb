W7A16 Cho một mảng số nguyên không âm. Gọi
mảng đó là "mảng đẹp" nếu mỗi phần tử
x trong mảng xuất hiện đúng x lần.
Kiểm tra xem mảng có phải là mảng đẹp
hay không. Nếu không hãy xóa đi y số (y
là nhỏ nhất có thể) để mảng còn lại thành
mảng đẹp.

```python
arr = list(map(int, input().split()))
counts = {}

for x in arr:
    counts[x] = counts.get(x, 0) + 1

dels = 0
for x, f in counts.items():
    if f >= x:
        dels += f - x
    else:
        dels += f

print(dels)
```

W7A17 Tìm phần tử xuất hiện nhiều hơn n/2 lần
trong mảng (nếu có) với độ phức tạp thời
gian là O(n), độ phức tạp bộ nhớ là O(1),
- Boyer–Moore. Nếu không có phần tử
nào thỏa mãn thì không in gì


```python
arr = list(map(int, input().split()))
candidate = None
count = 0

for num in arr:
    if count == 0:
        candidate = num
    
    if num == candidate:
        count += 1
    else:
        count -= 1

if arr.count(candidate) > len(arr) // 2:
    print(candidate)
```

W7A18 Sắp xếp có điều kiện: Số chẵn đứng
trước, số lẻ đứng sau, chẵn tăng dần, lẻ
giảm dần

```python
arr = list(map(int, input().split()))
evens = []
odds = []

for x in arr:
    if x % 2 == 0:
        evens.append(x)
    else:
        odds.append(x)

evens.sort()
odds.sort(reverse=True)

print(*(evens + odds))
```

W7A19 Merge 2 mảng đã sắp xếp (có thể tăng
dần, có thể giảm dần) mà không dùng
bất cứ thuật toán sắp xếp nào thành một
mảng cũng có cùng thứ tự sắp xếp như 2
mảng ban đầu

```python
a = list(map(int, input().split()))
b = list(map(int, input().split()))

tang_dan = True
if (len(a) > 1 and a[0] > a[-1]) or (len(b) > 1 and b[0] > b[-1]):
    tang_dan = False

if not tang_dan:
    a.reverse()
    b.reverse()

res = []
i = 0
j = 0

while i < len(a) and j < len(b):
    if a[i] < b[j]:
        res.append(a[i])
        i += 1
    else:
        res.append(b[j])
        j += 1

res.extend(a[i:])
res.extend(b[j:])

if not tang_dan:
    res.reverse()

print(*res)
```

W7A22 Cho N tên sinh viên gồm cả họ và tên,
tìm xem tên nào hay gặp nhất. Nếu có
nhiều tên cùng số lần gặp thì lấy tên có

W7A25 Tìm số có đúng k chữ số lớn nhất có thể
trong một số có n chữ số (coi đầu vào là
một số rất lớn - str). Không được hoán
đổi vị trí các chữ số


```python
s = input().strip()
k = int(input())

stack = []
can_remove = len(s) - k

for digit in s:
    while stack and can_remove > 0 and stack[-1] < digit:
        stack.pop()
        can_remove -= 1
    stack.append(digit)

print("".join(stack[:k]))
```

W7A26 Tách các số có trong một đoạn văn bản,
tính tổng và tính trung bình các số đó.

W7A27 Cho một dãy ngoặc ()(){}{}[][]..., kiểm tra
xem dãy ngoặc đã cho có hợp lệ hay
không

```python
s = input()
stack = []
hop_le = True

for c in s:
    if c in "({[":
        stack.append(c)
    elif c in ")}]":
        if not stack:
            hop_le = False
            break
        top = stack.pop()
        if (c == ')' and top != '(') or \
           (c == '}' and top != '{') or \
           (c == ']' and top != '['):
            hop_le = False
            break

if hop_le and not stack:
    print("True")
else:
    print("False")
```

W7A28 Bà
̀i toán Tìm dãy các hợp số dài nhất liên
tiếp không chứa số nguyên nào <= N (N
là số nguyên dương cho trước). Nếu có
nhiều dãy cùng độ dài thì lấy dãy nhỏ
nhất (gặp đầu tiên từ trái qua)

```python
n = int(input())
arr = list(map(int, input().split()))

def is_composite(k):
    if k < 2: return False
    for i in range(2, int(k**0.5) + 1):
        if k % i == 0: return True
    return False

max_len = 0
best_seq = []
curr_seq = []

for x in arr:
    if x > n and is_composite(x):
        curr_seq.append(x)
        if len(curr_seq) > max_len:
            max_len = len(curr_seq)
            best_seq = curr_seq[:]
    else:
        curr_seq = []

print(*best_seq)
```

W7A29 Tìm Dãy con tăng dần liên tục dài nhất
của một danh sách nhập vào từ bàn
phím. Lấy dãy đầu tiên nếu có nhiều dãy
có số phần tử bằng nhau

```python
arr = list(map(int, input().split()))
max_len = 0
best_seq = []
curr_seq = []

for x in arr:
    if not curr_seq or x > curr_seq[-1]:
        curr_seq.append(x)
    else:
        if len(curr_seq) > max_len:
            max_len = len(curr_seq)
            best_seq = curr_seq[:]
        curr_seq = [x]

if len(curr_seq) > max_len:
    best_seq = curr_seq

print(*best_seq)
```

W7A30 Tìm Dãy con tăng dần KHÔNG liên tục
dài nhất của một danh sách nhập vào từ
bàn phím

```python
arr = list(map(int, input().split()))
n = len(arr)

if n == 0:
    print()
else:
    dp = [1] * n
    parent = [-1] * n
    
    for i in range(1, n):
        for j in range(i):
            if arr[i] > arr[j] and dp[i] < dp[j] + 1:
                dp[i] = dp[j] + 1
                parent[i] = j
                
    max_len = max(dp)
    
    idx = -1
    for i in range(n):
        if dp[i] == max_len:
            idx = i
            break
            
    seq = []
    while idx != -1:
        seq.append(arr[idx])
        idx = parent[idx]
        
    seq.reverse()
    print(*seq)
```

W7A31 Đổi chỗ đúng 2 chữ số để được số nhỏ
nhất có thể

```python
s = list(input().strip())
n = len(s)
min_val = "".join(s)

for i in range(n):
    for j in range(i + 1, n):
        temp = s[:]
        temp[i], temp[j] = temp[j], temp[i]
        
        if temp[0] == '0' and n > 1:
            continue
            
        cand = "".join(temp)
        if cand < min_val:
            min_val = cand

print(min_val)
```
