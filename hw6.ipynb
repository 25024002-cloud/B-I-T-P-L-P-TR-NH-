W6A16
 vào mảng n hàng, m cột số nguyên. In lại mảng ra màn hình như ma ̣
trận hai chiều, định dạng mỗi phần tử 4 vị trí căn phải


```python
# 1. Nhập số hàng n và số cột m
n, m = map(int, input().split())

# 2. Nhập ma trận (List lồng List)
matrix = []
for i in range(n):
    # Mỗi hàng là một list các số nguyên
    hang = list(map(int, input().split()))
    matrix.append(hang)

# 3. In ma trận theo đúng định dạng
for hang in matrix:
    for so in hang:
        # :4d nghĩa là lấy 4 khoảng trống, số nguyên (d) sẽ nằm bên phải
        # end="" để các số trong cùng 1 hàng không bị xuống dòng
        print(f"{so:4d}", end="") 
    print()
```

W6A17
Nhập vào mảng vuông 2 chiều n hàng, n cột, in ra đường chéo chính và
đư

```python
n = int(input()) # Nhập kích thước n
matrix = []
for i in range(n):
    matrix.append(list(map(int, input().split())))

cheo_chinh = []
cheo_phu = []

for i in range(n):
    # Đường chéo chính: hàng i, cột i
    cheo_chinh.append(matrix[i][i])
    # Đường chéo phụ: hàng i, cột n-1-i
    cheo_phu.append(matrix[i][n-1-i])

print(cheo_chinh)
print(cheo_phu)
```

W6A18 Nhập vào mảng n hàng, m cột số nguyên và số nguyên k (n, m, k nhập trên ̣
cùng một dòng, cách nhau dấu cách). Tính và in ra tổng các phần tử trên cột
k (hàng k) 


```python
# Nhập n, m, k trên cùng 1 dòng
n, m, k = map(int, input().split())

matrix = []
for i in range(n):
    matrix.append(list(map(int, input().split())))

tong_cot = 0
# Duyệt qua từng hàng, nhưng chỉ lấy số ở đúng cột k
for i in range(n):
    tong_cot += matrix[i][k] 

print(tong_cot)
```

W6A19 Viết hàm đưa vào 1 list số nguyên, tìm và trả về vị trí có giá trị lớn nhất
trong list

```python
def vi_tri_max(lst):
    vi_tri_max = 0
    for i in range(1, len(lst)):
        if lst[i]> lst[vi_tri_max]:
            vi_tri_max = i
    return vi_tri_max

lst = list(map(int, input().split()))
print(vi_tri_max(lst))
```

W6A23 VIết hàm xóa đi các số <x có trong mảng

```python
def xoa(lst1, x):
    lst2 = []
    for i in lst1:
        if i >= x:
            lst2.append(i)
    return lst2
lst1 = map(int, input().split())
x = int(input()) 
print(xoa(lst1, x))
```

W6A24 Viết hàm đẩy tất cả các số 0 về đầu (về cuối), các số khác vẫn đúng vị trí
trước sau: 0 2 3 2 4 0 0 0 9 => 0 0 0 2 3 2 4 9



```python
def day_len(lst):
    lst0 = []
    lst_bth = []
    for i in lst:
        if i == 0:
            lst0.append(i)
        else:
            lst_bth.append(i)
    return lst0 + lst_bth

lst = list(map(int, input().split()))
print(day_len(lst))
            
```

W6A26 Tính tổng các phần tử cực đại địa phương trong một danh sách, không tính 2
phần tử đầu và cuối

```python
arr = list(map(int, input().split()))
tong = 0
# Bỏ thằng đầu (vị trí 0) và thằng cuối (vị trí len-1)
for i in range(1, len(arr) - 1):
    if arr[i] > arr[i-1] and arr[i] > arr[i+1]:
        tong = tong + arr[i]

print(tong)
```

W6A27 Cho một dãy số nguyên, Tìm và in ra vị trí các số < x (x là giá trị cho trước).
nếu không có in ra -1


```python
x = int(input())
nums = list(map(int, input().split()))
vitri = []
for i in range(len(nums)):
    if nums[i] < x:
        vitri.append(i)
if len(vitri) == 0:
    print("-1")
else:
    for v in vitri:
        print(v, end=" ")
```

W6A28 Cho một dãy các số nguyên, kiểm tra xem nó là dãy tăng thực sự hay giảm
thực sự, hay vô danh (không tăng, không giảm)

```python
nums = list(map(int, input().split()))
is_tang = all(nums[i] < nums[i+1] for i in range(len(nums) - 1))
is_giam = all(nums[i] > nums[i+1] for i in range(len(nums) - 1))

if is_tang:
    print("tang thuc su")
elif is_giam:
    print("giam thuc su")
else:
    print("vo danh")
```
