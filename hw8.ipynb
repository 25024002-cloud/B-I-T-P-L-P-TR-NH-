Xây dựng lớp Account(balance) với hai phương thức deposit(x) và get_balance().
Xây dựng lớp SavingAccount(balance, rate) kế thừa từ Account, thêm phương thức
add_interest() để cộng balance*rate vào số dư.
Input
balance rate
Output
Số dư sau khi cộng lãi, in 2 chữ số thập phân. 

```python
class Account:
    def __init__(self, balance):
        self.balance = balance
        
    def deposit(self, x):
        self.balance += x
        
    def get_balance(self):
        return self.balance

class SavingAccount(Account):
    def __init__(self, balance, rate):
        super().__init__(balance)
        self.rate = rate
        
    def add_interest(self):
        self.balance += self.balance * self.rate

du_lieu = input().split()
balance = float(du_lieu[0])
rate = float(du_lieu[1])

acc = SavingAccount(balance, rate)
acc.add_interest()
print(f"{acc.get_balance():.2f}")
```

W8A22 Yêu cầu
Xây dựng lớp Matrix2x2(a, b, c, d) biểu diễn ma trận:
[a b]
[c d]
Viết phương thức trace() trả về a + d và determinant() trả về ad - bc.
Input
Một dòng chứa 4 số nguyên a b c d.
Output
In ra trace và determinant trên cùng một dòng. 


```python
class Matrix2x2:
    def __init__(self, a, b, c, d):
        self.a = a
        self.b = b
        self.c = c
        self.d = d
        
    def trace(self):
        return self.a + self.d
        
    def determinant(self):
        return self.a * self.d - self.b * self.c

a, b, c, d = map(int, input().split())
m = Matrix2x2(a, b, c, d)
print(f"{m.trace()} {m.determinant()}")
```

W8A25 Yêu cầu
Xây dựng lớp PolynomialSum cho hai đa thức bậc nhất:
P(x) = ax + b, Q(x) = cx + d.
Viết phương thức sum_str() trả về chuỗi biểu diễn đa thức tổng dưới dạng kx + m.
Input
a b c d
Output
Chuỗi của đa thức P(x)+Q(x). 

```python
class PolynomialSum:
    def __init__(self, a, b, c, d):
        self.a = a
        self.b = b
        self.c = c
        self.d = d
        
    def sum_str(self):
        k = self.a + self.c
        m = self.b + self.d
        
        if k.is_integer(): k = int(k)
        if m.is_integer(): m = int(m)
        
        if m < 0:
            return f"{k}x - {abs(m)}"
        else:
            return f"{k}x + {m}"

du_lieu = input().split()
a = float(du_lieu[0])
b = float(du_lieu[1])
c = float(du_lieu[2])
d = float(du_lieu[3])

poly_sum = PolynomialSum(a, b, c, d)
print(poly_sum.sum_str())
```

W8A27 Yêu cầu
Xây dựng lớp Product(name, price).
Xây dựng lớp DiscountProduct(name, price, percent) kế thừa từ Product, ghi đè phương thức
final_price() để trả về giá sau giảm.
Công thức: price * (100 - percent) / 100
Input
name price percent
Output
In ra giá cuối cùng với 2 chữ số thập phân. 

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price
        
    def final_price(self):
        return self.price

class DiscountProduct(Product):
    def __init__(self, name, price, percent):
        super().__init__(name, price)
        self.percent = percent
        
    def final_price(self):
        return self.price * (100 - self.percent) / 100

du_lieu = input().split()
name = du_lieu[0]
price = float(du_lieu[1])
percent = float(du_lieu[2])

dp = DiscountProduct(name, price, percent)
print(f"{dp.final_price():.2f}")
```

W8A28 Yêu cầu
Xây dựng lớp Counter có thuộc tính count ban đầu bằng 0 và các phương thức inc(), dec(),
value().
Sau đó thực hiện dãy thao tác trên đối tượng.
Input
Dòng 1: n
Dòng 2: n thao tác, mỗi thao tác là INC hoặc DEC, cách nhau bởi dấu cách.
Output
Giá trị cuối cùng của count

```python
class Counter:
    def __init__(self):
        self.count = 0
        
    def inc(self):
        self.count += 1
        
    def dec(self):
        self.count -= 1
        
    def value(self):
        return self.count

n = int(input())
ops = input().split()

c = Counter()
for op in ops:
    if op == "INC":
        c.inc()
    elif op == "DEC":
        c.dec()

print(c.value())
```

W8A29 Yêu cầu
Xây dựng lớp Time(h, m, s) với phương thức to_seconds() đổi thời gian sang tổng số giây.
Input
h m s
Output
Tổng số giây

```python
class Time:
    def __init__(self, h, m, s):
        self.h = h
        self.m = m
        self.s = s
        
    def to_seconds(self):
        return self.h * 3600 + self.m * 60 + self.s

h, m, s = map(int, input().split())
t = Time(h, m, s)
print(t.to_seconds())
```

W8A30 Yêu cầu
Xây dựng lớp Stack sử dụng danh sách để cài đặt ngăn xếp với các phương thức push(x),
pop(), top().
Dữ liệu đảm bảo mọi lệnh pop và top đều hợp lệ.
Input
Dòng 1: n
n dòng tiếp theo là một trong ba lệnh:
PUSH x
POP
TOP
Output
Mỗi lệnh TOP in ra giá trị ở đỉnh ngăn xếp trên một dòng

```python
class Stack:
    def __init__(self):
        self.items = []
        
    def push(self, x):
        self.items.append(x)
        
    def pop(self):
        self.items.pop()
        
    def top(self):
        return self.items[-1]

n = int(input())
s = Stack()

for _ in range(n):
    lenh = input().split()
    if lenh[0] == "PUSH":
        s.push(int(lenh[1]))
    elif lenh[0] == "POP":
        s.pop()
    elif lenh[0] == "TOP":
        print(s.top())
```
